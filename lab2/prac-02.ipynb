```python
import numpy as np 
import cv2
import matplotlib.pyplot as plt  
 
```

```python
image = cv2.imread("dog.jpg", cv2.IMREAD_GRAYSCALE)
plt.imshow(image, cmap="gray")
plt.show()
```

```python
def add_salt_pepper_noise(image, probability=0.3):
    noisy = image.copy()

    random = np.random.rand(*image.shape)

    noisy[random < probability / 2] = 0      # Pepper noise
    noisy[random > 1 - probability / 2] = 255  # Salt noise

    return noisy
```

```python
prob = 0.5
noisy = add_salt_pepper_noise(image,prob)
plt.imshow(noisy,cmap='gray')
plt.show
```
